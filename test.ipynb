```python
import websockets
import json

HASS_WS_URL = "ws://192.168.0.180:8123/api/websocket"
with open(".secrets/hass.txt") as f: 
    TOKEN = f.read()

async def connect_to_hass():
    async with websockets.connect(HASS_WS_URL) as ws:
        # Step 1: Receive initial message from HASS
        hello = await ws.recv()
        print("Received hello:", hello)

        # Step 2: Send authentication message
        await ws.send(json.dumps({
            "type": "auth",
            "access_token": TOKEN
        }))

        # Step 3: Wait for auth response
        auth_response = await ws.recv()
        print("Auth response:", auth_response)

        # Step 4: Send request to get entity state
        await ws.send(json.dumps({
            "id": 1,
            "type": "get_states"
        }))

        # Step 5: Receive states
        states_response = await ws.recv()
        states = json.loads(states_response)
        print("States received:", states)

        # Optional: filter for a specific entity
        for entity in states.get("result", []):
            if entity["entity_id"] == "light.living_room":
                print(f"Living room light state: {entity['state']}")

# Run the async function
await connect_to_hass()
```
